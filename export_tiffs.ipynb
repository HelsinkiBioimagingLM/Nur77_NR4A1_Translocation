```python
from bioio import BioImage
import bioio_lif
import os
import re
import napari
from glob import glob
from bioio.writers import OmeTiffWriter #not a good fit for img data that are not matching in all list lengths for metadata
import numpy as np
import dask as dk
from skimage.io import imsave
```

```python
files = sorted(glob("/DISKS/2TB/gkristin/Nur77_NR4A1_Translocation/LIF_Data/*.lif"))
```

```python
names = list(map(os.path.basename,files))
```

```python
for name, file in zip(names,files):
    img_meta = BioImage(file, reader=bioio_lif.Reader)
    for i, scene in enumerate(img_meta.scenes):
        new_name = name.replace('.lif',f"_scene_{i}.tif")
        img_meta.set_scene(i) #use to switch between scenes/positiions in multi-scene/position data
        img = np.squeeze(img_meta.get_image_data())
        imsave(f"/DISKS/2TB/gkristin/Nur77_NR4A1_Translocation/Exported_Tiffs/{new_name}",img,check_contrast=False)
```
